```python
#Q3: Who are the top 10 customers by total account balance across all account types?

queries = {
"Q3": """SELECT c.customer_id, c.name, 
            SUM(a.account_balance) AS total_balance
            FROM customers c 
            JOIN accounts a ON a.customer_id = c.customer_id
            GROUP BY c.customer_id, c.name, c.account_type
            ORDER BY total_balance DESC LIMIT 10;"""
}
```

```python
#Q4: Which customers opened accounts in 2023 with a balance above ₹1,00,000?

queries = {
"Q4": """SELECT c.customer_id, c.name, c.city, c.account_type, a.account_balance, c.join_date
            FROM customers c
            JOIN accounts a 
            ON c.customer_id = a.customer_id
            WHERE strftime('%Y', c.join_date) = '2023' 
            AND a.account_balance > 100000
            ORDER BY a.account_balance DESC LIMIT 3;"""
}
```

```python
for q , sql in queries.items():
    print(f"\n {q}:")
    try:
        df = pd.read_sql_query(sql, conn)
        print(df.head(10))
    except Exception as e:
        print(f" Error in {q}:{e}")
 
```

```python
#Q5: What is the total transaction volume (sum of amounts) by transaction type?
queries = {
"Q5":  """SELECT txn_type, 
            SUM(transactions.amount) AS total_amount
            FROM transactions
            GROUP BY txn_type
            ORDER BY total_amount DESC ;"""
}
```

```python
#Q6: Which accounts have more than 3 failed transactions in a single month?
queries = {
"Q6": """SELECT customer_id,
            strftime('%y-%u',txn_time) AS txn_month,
            COUNT(*) AS failed_txn_count
            FROM transactions
            WHERE status = 'failed' 
            GROUP BY customer_id,txn_month
            HAVING COUNT(*)> 3
            ORDER BY failed_txn_count DESC LIMIT 3;"""
}
```

```python
for q , sql in queries.items():
    print(f"\n {q}:")
    try:
        df = pd.read_sql_query(sql, conn)
        print(df.head(10))
    except Exception as e:
        print(f" Error in {q}:{e}")
 
```

```python
#Q7: Which are the top 5 branches by total transaction volume in the last 6 months?
queries = {
"Q7": """SELECT b.branch_id, b.branch_name,
            SUM(t.amount) AS total_volume
            FROM transactions t
            JOIN branches b ON t.customer_id = b.branch_id
            WHERE txn_time >= DATE('now','-6 months')
            GROUP BY b.branch_id, b.branch_name
            ORDER BY total_volume DESC LIMIT 5;"""
}

```

```python
for q , sql in queries.items():
    print(f"\n {q}:")
    try:
        df = pd.read_sql_query(sql, conn)
        print(df.head(10))
    except Exception as e:
        print(f" Error in {q}:{e}")
```

```python
#Q13: What is the branch performance summary showing total customers, total loans, and transaction volume?
queries = {
"Q13": """SELECT b.Branch_Name,
            COUNT(DISTINCT a.customer_id) AS total_customers,
            COUNT(DISTINCT l.Loan_ID) AS total_loans,
            ROUND(SUM(a.account_balance),2) AS total_balance
            FROM branches b
            LEFT JOIN accounts a ON b.Branch_ID = a.customer_id
            LEFT JOIN loans l ON a.customer_id  = l.Customer_ID 
            GROUP BY b.Branch_Name
            ORDER BY total_balance DESC LIMIT 10;"""
}
```

```python
for q , sql in queries.items():
    print(f"\n {q}:")
    try:
        df = pd.read_sql_query(sql, conn)
        print(df.head(10))
    except Exception as e:
        print(f" Error in {q}:{e}")
```

```python
#Q14: Which issue categories have the longest average resolution time?
queries = {
"Q14": """SELECT issue_category,
            ROUND(AVG(julianday(Date_Closed)-julianday(Date_Opened)),2) AS avg_days, COUNT(*) AS total_tickets
            FROM support_tickets
            WHERE Date_Closed!=''
            GROUP BY issue_category
            ORDER BY avg_days DESC LIMIT 10;"""
}
```
